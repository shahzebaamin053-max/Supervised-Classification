```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

print("Libraries import Successfully.")
```

```python
df = pd.read_csv('/content/Housing.csv', on_bad_lines='skip', low_memory=False)
print("Data Loaded successfully.")
df.head()
```

```python
from sklearn.linear_model import Ridge

# define target and features
target = 'price'

# Drop rows with any missing values before defining X and y
df.dropna(inplace=True)

# separate featuers X and y target
X = df.drop(columns=[target, 'id', 'date'])
y = df[target]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 40)

ridge_model = Ridge(alpha = 1.0)
print("Ridge Regression model initialized.")

ridge_model.fit(X_train, y_train)
print("Ridge Regression model trained successfully.")

y_pred_ridge = ridge_model.predict(X_test)
print("Predictions made on the test set.")

plt.figure(figsize = (10, 6))
plt.scatter(y_test, y_pred_ridge, alpha= 0.7, color = 'blue', label = 'Predicted vs. actual')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2, label = 'Perfect prediction line')
plt.xlabel('Actual House Prices ($1000s)')
plt.ylabel('Predicted House Prices ($1000s)')
plt.title('Actual vs. Predicted House Prices')
plt.legend()
plt.grid(True)
plt.show()
```
